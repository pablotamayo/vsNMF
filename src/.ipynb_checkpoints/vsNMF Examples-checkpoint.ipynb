```python
CRISPR_file = io.StringIO(CRISPR_data)
RNAi_file = io.StringIO(RNAi_data)
CTRP_file = io.StringIO(CTRP_data)
```

```python
CRISPR = pd.read_csv(CRISPR_file, index_col=["Gene"])
RNAi = pd.read_csv(RNAi_file, index_col=["Gene"])
CTRP = pd.read_csv(CTRP_file, index_col=["Gene"])
```

## Assemble the 3 datasets as a 3D array and convert it into a tensor with the following dimensions: perturbation type (CRISPR, RNAi, Drug), Gene and Cell Line

The drugs are identified by their target gene, The data correspond to compounds: 'dabrafenib (inhibitor of BRAF)', 'selumetinib (inhibitor of MEK1 and MEK2)','erlotinib (inhibitor of EGFR and HER2)','imatinib (inhibitor of BCR-ABL1 and c-KIT)','AZD6482 (inhibitor of PI3K catalytic subunits beta and delta)','crizotinib (inhibitor of c-MET and ALK)','BMS-536924 (inhibitor of insulin-like growth factor 1 receptor (IGF1R) and insulin receptor (INSR))','saracatinib (inhibitor of SRC and ABL1)','BRD-K55473186 (stapled helical peptide targeting NOTCH1)'

```python
array_3D = np.stack([CRISPR.to_numpy(), RNAi.to_numpy(), CTRP.to_numpy()], axis=0)
array_3D
```

```python
input_tensor = tl.tensor(array_3D, dtype="float")
```

## Define sparsity constraints for the 3 factor matrices

```python
sparsity_coefficients = [[0.6, 0.7, 0.7], [0.2, 0.3, 0.4], [0.4, 0.5, 0.6]]
```

## Decompose the input tensor using vsNTF

```python
from vsNTF.vsNTF import vsNTF

core, factors = vsNTF(input_tensor, rank=[3, 3, 3], return_errors=False,
    #tensor,
    #rank,
    n_iter_max = 1000,
    init ='random',
    tol = 0.0001,
    sparsity_coefficients = sparsity_coefficients,
    normalize_initial_tensor_input = True,
    random_state = None,
    verbose = False,
    normalize_factors = False,
)
```

```python
core
```

```python
factors
```

```python
factor_A = pd.DataFrame(factors[0], index = ['CRISPR', 'RNAi', 'CTRP'], 
                        columns = ['F{}'.format(x) for x in range(factors[0].shape[1])])

factor_B = pd.DataFrame(factors[1], index = CRISPR.index, 
                        columns = ['F{}'.format(x) for x in range(factors[1].shape[1])])

factor_C = pd.DataFrame(factors[2], index = CRISPR.columns, 
                        columns = ['F{}'.format(x) for x in range(factors[2].shape[1])])
```

```python
factor_C
```

## Display factor matrices sparseness

```python
from vsNMF.vsNMF import compute_row_sparseness

factor_A_vec, factor_A_aver_sparseness = compute_row_sparseness(factor_A.T)
factor_B_vec, factor_B_aver_sparseness = compute_row_sparseness(factor_B.T)
factor_C_vec, factor_C_aver_sparseness = compute_row_sparseness(factor_C.T)

print(factor_A_vec)
print(factor_B_vec)
print(factor_C_vec)
```

## Display the core and factor matrices

```python
from seaborn import heatmap, clustermap
from matplotlib.pyplot import get_cmap
from vsNMF import normalize_dataset_vs_reference

factor_A_norm = normalize_dataset_vs_reference(factor_A, factor_A, clims=[-3, 3])
factor_A_norm.index.name = 'Perturbations'
factor_A_norm.columns.name = 'Signatures'
factor_A_norm.columns = ['{} (sparsity: {})'.format(x, y) for x, y in zip(factor_A_norm.columns, 
                                                                          np.round(factor_A_vec, 2))]

clustermap(factor_A_norm.T, cmap = get_cmap('bwr'), figsize=(10, 5), row_cluster = False,
                linewidths = 0, linecolor = 'black', robust = False, cbar_pos = None,
                annot=False, cbar = False, square = False, yticklabels = 'auto', mask = None)
```

```python
factor_B_norm = normalize_dataset_vs_reference(factor_B, factor_B, clims=[-3, 3])
factor_B_norm.index.name = 'Genes'
factor_B_norm.columns.name = 'Signatures'
factor_B_norm.columns = ['{} (sparsity: {})'.format(x, y) for x, y in zip(factor_B_norm.columns, 
                                                                          np.round(factor_B_vec, 2))]

clustermap(factor_B_norm.T, cmap = get_cmap('bwr'), figsize=(10, 5), row_cluster = False,
                linewidths = 0, linecolor = 'black', robust = False, cbar_pos = None,
                annot=False, cbar = False, square = False, yticklabels = 'auto', mask = None)
```

```python
factor_C_norm = normalize_dataset_vs_reference(factor_C, factor_C, clims=[-3, 3])
factor_C_norm.index.name = 'Cell Lines'
factor_C_norm.columns.name = 'Signatures'
factor_C_norm.columns = ['{} (sparsity: {})'.format(x, y) for x, y in zip(factor_C_norm.columns, 
                                                                          np.round(factor_C_vec, 2))]

clustermap(factor_C_norm.T, cmap = get_cmap('bwr'), figsize=(10, 5), row_cluster = False,
                linewidths = 0, linecolor = 'black', robust = False, cbar_pos = None,
                annot=False, cbar = False, square = False, yticklabels = 'auto', mask = None)
```
